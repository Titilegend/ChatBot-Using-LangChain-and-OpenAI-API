```python
import os
from google.colab import userdata
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

if not os.environ["OPENAI_API_KEY"]:
    raise ValueError("OPENAI_API_KEY is not found")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Answer clearly and concisely"),
    ("user","{question}")
])


llm = ChatOpenAI(model = "gpt-4o-mini",temperature=0.2)

chain = prompt | llm | StrOutputParser()
```

```python
tests = [

         "What is LangChain in one paragraph?",
         "Explain Government like I'm 5 years old",
         "Recommend 5 beginner ML Projects using python",
         "Write a short email requesting an email",
         "Difference between SQL nad MySQL?"

]

for i,q in enumerate(tests,1):
  a = chain.invoke({"question": q})
  print(f"Q{i}: {q}\nA{i}: {a}\n" + "-"*60)
```

```python
while True:
  q = input("You: ").strip()
  if q.lower() in {"exit","quit"}:
    print("Bot: Goodbye")
    break
  if not q:
    continue
  print("Bot",chain.invoke({"question": q}), "\n")
```
